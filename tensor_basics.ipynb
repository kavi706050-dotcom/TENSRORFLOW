## What is a tensor?

A tensor is just a **box of numbers**.

That is really all it is. The only difference between tensors is how many
directions you can move inside the box.

| Name | Example | Picture |
| --- | --- | --- |
| Scalar | `5` | a single dot |
| Vector | `[1, 2, 3]` | a line |
| Matrix | `[[1, 2], [3, 4]]` | a table |
| 3D tensor | many tables stacked | a cube |

The word "tensor" in TensorFlow comes from this. Data flows through your model as tensors.

```python
import tensorflow as tf
```

```python
scalar = tf.constant(5)

scalar
```

```python
vector = tf.constant([1,2,3])

vector
```

```python
matrix = tf.constant([[1,2],[3,4]])

matrix
```

```python
cube = tf.constant([[[1,2],[3,4]],[[5,6],[7,8]]])

cube
```

## Shape and rank

- **shape** tells you how many items in each direction
- **rank** tells you how many directions there are

Rank is simply how many numbers are in the shape.

```python
print(scalar.shape, tf.rank(scalar).numpy())
print(cube.shape, tf.rank(cube).numpy())
print(matrix.shape, tf.rank(matrix).numpy())
```

```python
a = tf.constant([1,2,3])
b = tf.constant([10,20,30])

print((a + b).numpy())
print(a - b)
print(a * b)
print(a / b)
```

```python
m1 = tf.constant([[1,2],[3,4]])

m2 = tf.constant([[1,0],[0,1]])
```

```python
print( (m1 * m2).numpy())
```
