## What is TensorFlow?

TensorFlow is a free tool made by Google that helps you build machine learning and deep learning models.

Think of it like this:

- You give it some **data** (examples)
- You tell it what **answer** you want
- TensorFlow finds the **pattern** for you

You do not write the rules yourself. The computer learns the rules from examples.

## A simple picture

Normal programming:

```
data + rules  ->  computer  ->  answers
```

Machine learning & Deep learning:

```
data + answers  ->  computer  ->  rules
```

TensorFlow is the tool that finds those rules.

## Every TensorFlow program has 3 steps

1. **Build** the model (choose the layers)
2. **Compile** the model (choose how it learns)
3. **Train** the model (show it the examples)

That is it. Let us do all 3 right now.

````python
import numpy as np

x = np.array([1,2,3,4,5,6], dtype=float)
y = np.array([3,5,7,9,11,13], dtype=float)

#y = w * x + b
````

````python
model = keras.Sequential([
    keras.layers.Input(shape=(1,)),
    keras.layers.Dense(1)
])
````

````python

model.compile(optimizer="sgd", loss="mean_squared_error")
````

````python
model.fit(x,y, epochs=500, verbose=0)

print("Training Done")
````

````python
answer = model.predict(np.array([10.0]), verbose=0)

print("Model Says",answer)
print("correct answer is 21")
````

````python
weight, bias = model.layers[0].get_weights()

print(weight)
print(bias)
````
