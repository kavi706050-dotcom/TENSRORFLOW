## Joining tensors

- `tf.concat` glues tensors together, shape stays the same rank
- `tf.stack` puts them on top of each other, rank goes up by one

```python
a = tf.constant([[1,2],[3,4]])

b = tf.constant([[5,6],[7,8]])

```

```python
print(tf.concat([a,b], axis=0).numpy())
```

```python
stacked = tf.stack([a,b])

print(stacked.shape)
print(stacked)
```
